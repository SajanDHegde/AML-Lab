```python
""" 
3. Create the following DataFrame:

| Name  | Age | Department | Marks |
| ----- | --: | ---------- | ----: |
| Asha  |  22 | CSE        |    85 |
| Ravi  |  23 | ECE        |    78 |
| Meena |  21 | CSE        |    92 |
| Kiran |  24 | ME         |    74 |
| Arun  |  22 | ECE        |    88 |

Display:

	Complete DataFrame
	First 3 rows
	Last 2 rows
	Shape
	Column names
	Data types
"""

data = {
    'Name':['Asha', 'Ravi', 'Meena', 'Kiran', 'Arun'],
    'Age': [22,23,21,24,22],
    'Department':['CSE', 'ECE', 'CSE', 'ME', 'ECE'],
    'Marks':[85,78,92,74,88]
}

students = pd.DataFrame(data)
print("Complete DataFame\n",students)
print('\nFirst 3 rows\n',students.head(3))
print('\nLast 2 rows\n',students.tail(2))
print('\nShape\n',students.shape)
print('\nColumns\n',students.columns)
print('\nData Types\n',students.dtypes)
```

```python
''' 
4. Using the previous DataFrame, display:

	info()
	describe()
	Number of rows
	Number of columns
'''
print('\nFifo\n',students.info())
print('\ndescribe\n',students.describe())
print('\nNumbers of rows\n',students.shape[0])
print('\nNumbers of columns\n',students.shape[1])
```

```python
''' 
5. Using df, display:

	Name column
	Marks column
	Name and Marks together
'''
print('\nName Column\n',students['Name'])
print('\nMarks Column\n',students['Marks'])
print('\nName and Marks Column\n',students[['Name','Marks']])
```

```python
''' 
6. Display:

	First row using iloc
	Third row using iloc
	Rows 1 to 3
	Name and Marks columns using loc
'''

print('\nFirst row using iloc\n',students.iloc[0])
print('\nThird row using iloc\n',students.iloc[3])
print('\nFirst to Third rows using iloc\n',students.iloc[:3])
print('\nName and Marks columns using loc\n',students.loc[:, ['Name', 'Marks']])
```

```python
''' 
7. Find:

	Students scoring above 80.
	Students scoring between 75 and 90.
	Students belonging to CSE.
	CSE students scoring above 80.
'''
print('\nStudents scoring above 80: \n',students[students['Marks']>80])
print(
    '\nStudents scoring between 75 and 90:\n',students[students['Marks'].between(75, 90)]
)
print('\nStudents belonging to CSE: \n',students[students['Department']=='CSE'])
print('\nCSE students scoring above 80: \n',students[(students['Department']=='CSE') & (students['Marks']>80)])
```

```python
''' 
8. Display students belonging to either CSE or ECE.
'''
print('\nDisplay students belonging to either CSE or ECE:\n',
    students[students['Department'].isin(['CSE', 'ECE'])])
```

```python
''' 
9. Add a new column called Result. Students scoring 75 or above should receive "Pass", otherwise "Fail".
'''
# np.where -> used to create a new column based on a condition. 
# It takes three arguments: the condition, the value to assign if the condition is True,
#  and the value to assign if the condition is False.
students['Result'] = np.where(
    students['Marks'] >= 75,
    'Pass',
    'Fail'
)

print(students) 
```

```python
''' 
10. Assign grades according to:

	| Marks | Grade |
	| ----: | ----- |
	|  ≥ 90 | A     |
	|  ≥ 80 | B     |
	|  ≥ 70 | C     |
	|  < 70 | D     |

'''
# pd.cut -> used to segment and sort data values into bins.
bins = [0, 70, 80, 90, 100]
labels = ['D', 'C', 'B', 'A']
students['Grade'] = pd.cut(students['Marks'], bins=bins, labels=labels, right=False)
print(students)
```

```python
''' 
11. Sort students:

	By marks in ascending order.
	By marks in descending order.
	By Department and then Marks.
'''
print('\nBy marks in ascending order:\n',students.sort_values('Marks'))
print('\nBy marks in descending order:\n',students.sort_values('Marks', ascending=False))
print('\nBy Department and then Marks:\n',students.sort_values(['Department', 'Marks']))
```

```python
''' 
12. Consider:

data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
    "Age": [22, 23, np.nan, 24, 22],
    "Marks": [85, np.nan, 92, 74, 88]
}

students = pd.DataFrame(data)

Find:

Missing values
Number of missing values in each column
Rows containing missing values

'''

data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
    "Age": [22, 23, np.nan, 24, 22],
    "Marks": [85, np.nan, 92, 74, 88]
}

students = pd.DataFrame(data)

print('\nMissing values:\n',students.isnull())
print('\nNumber of missing values in each column:\n',students.isnull().sum())
print('\nRows containing missing values:\n',students[students.isnull().any(axis=1)])
# any() -> used to check if any element in a DataFrame or Series is True along a specified axis.
```

```python
''' 
13. Replace:

	Missing Age with average Age
	Missing Marks with average Marks
'''
x = students.copy()

x['Age'] = x['Age'].fillna(students['Age'].mean())
x['Marks'] = x['Marks'].fillna(students['Marks'].mean())
print('\nAfter replacing missing values:\n', x)
```

```python
''' 
16. Given:

marks = pd.DataFrame({
    "Student": ["A", "B", "C", "D", "E"],
    "Marks": [78, 85, 92, 67, 88]
})

Calculate:

Mean
Median
Maximum
Minimum
Standard deviation
Sum

'''

marks = pd.DataFrame({
    "Student": ["A", "B", "C", "D", "E"],
    "Marks": [78, 85, 92, 67, 88]
})

print('\nMean:\n',marks['Marks'].mean())
print('\nMedian:\n',marks['Marks'].median())    
print('\nMaximum:\n',marks['Marks'].max())
print('\nMinimum:\n',marks['Marks'].min())
print('\nStandard deviation:\n',marks['Marks'].std())
print('\nSum:\n',marks['Marks'].sum())
```

```python
''' 
17. Consider:

data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE"
    ],

    "Marks": [
        85, 78, 92,
        74, 88, 80
    ]
}

df = pd.DataFrame(data)


Find:
Average marks by department
Maximum marks by department
Number of students in each department

For each department, calculate:
Mean
Minimum
Maximum
Count

'''
data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE"
    ],

    "Marks": [
        85, 78, 92,
        74, 88, 80
    ]
}

df = pd.DataFrame(data)

x = df.groupby('Department')
print('\nAverage marks by department:\n',x['Marks'].mean())
print('\nMaximum marks by department:\n',x['Marks'].max())
print('\nNumber of students in each department:\n',x.size())
print('\nFor each department, calculate:\n',x['Marks'].agg(['mean', 'min', 'max', 'count']))
```

```python
''' 
20. Create:

employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],

    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],

    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],

    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})


Determine:

Average salary.
Maximum and minimum salary.
Employee receiving the highest salary.
Employees earning above ₹60,000.
Average salary for each department.
Average experience by department.
Number of employees in each department.
Sort employees by salary.
Find IT employees with more than 3 years of experience.
Add a SalaryLevel column:
	≥ ₹75,000 → High
	₹50,000–₹74,999 → Medium
	< ₹50,000 → Low
'''

employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],

    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],

    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],

    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})

print('\nAverage salary:\n',employees['Salary'].mean())
print('\nMaximum salary:\n',employees['Salary'].max())
print('\nMinimum salary:\n',employees['Salary'].min())
print('\nEmployee receiving the highest salary:\n',employees.loc[employees['Salary'].idxmax()])
print('\nEmployees earning above ₹60,000:\n',employees[employees['Salary'] > 60000])
dep = employees.groupby('Department')
print('\nAverage salary for each department:\n',dep['Salary'].mean())
print('\nAverage experience by department:\n',dep['Experience'].mean())
print('\nNumber of employees in each department:\n',dep.size())
print('\nEmployees sorted by salary:\n',employees.sort_values('Salary')) 
print('\nIT employees with more than 3 years of experience:\n',employees[(employees['Department']=='IT') & (employees['Experience']>3)])
bins = [0, 50000, 75000, np.inf]
labels = ['Low', 'Medium', 'High']
employees['SalaryLevel'] = pd.cut(employees['Salary'], bins=bins, labels=labels, right=False)
print('\nEmployees with SalaryLevel:\n',employees)	

```
