```python
""" 
9. Given: marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84]) 
 
Using NumPy boolean indexing, find: 
 
Marks greater than 75 
Marks between 50 and 80 
Number of students who scored below 50 
Percentage of students who scored 60 or above 
 
Do not use a Python for loop. 
"""
marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])
print("Marks greater than 75:", marks[marks > 75])
print("Marks between 50 and 80:", marks[(marks >= 50) & (marks <= 80)])
print("Number of students who scored below 50:", np.sum(marks < 50))
print("Percentage of students who scored 60 or above:", (np.sum(marks >= 60) / len(marks)) * 100)
```

```python
""" 
10. The following array contains marks of four students in three 
subjects: 
 
marks = np.array([ 
    [80, 75, 90], 
    [65, 70, 72], 
    [88, 92, 85], 
    [55, 60, 58] 
]) 
 
Calculate: 
 
Total marks of each student 
Average marks of each student 
Average marks in each subject 
Student with the highest total 
Highest mark in each subject. 
"""
marks = np.array([
    [80, 75, 90],
    [65, 70, 72],
    [88, 92, 85],
    [55, 60, 58]
])
print("Total marks of each student:", marks.sum(axis=1))
print("Average marks of each student:", marks.mean(axis=1))
print("Average marks in each subject:", marks.mean(axis=0))
print("Student with the highest total:", np.argmax(marks.sum(axis=1)))
print("Highest mark in each subject:", marks.max(axis=0))
```

```python
""" 
11. Create: 
 
A = np.array([ 
    [1, 2], 
    [3, 4] 
]) 
 
B = np.array([ 
    [5, 6], 
    [7, 8]
    ]) 
 
Perform: 
 
Matrix addition 
Element-wise multiplication 
Matrix multiplication 
Transpose of A. 
 
explain the difference between: 
A * B and A @ B 
 
"""

A = np.array([
    [1, 2], 
    [3, 4] ])

B = np.array([
    [5, 6], 
    [7, 8]
])  

print("Matrix addition:\n", A + B)
print("Element-wise multiplication:\n", A * B)
print("Matrix multiplication:\n", A @ B)
print("Transpose of A:\n", A.T)
print()

print("A*B is element-wise multiplication, where each element of A is multiplied by the corresponding element of B. \nWhereas A@B is matrix multiplication, which involves taking the dot product of rows of A with columns of B.")
```

```python
""" 
12. Daily temperatures for two weeks are: 
 
temperature = np.array([ 
    31, 32, 33, 35, 36, 34, 32, 
    30, 29, 31, 33, 35, 37, 36 
]) 
 
Find: 
 
Mean temperature 
Maximum and minimum temperatures 
Number of days above the average temperature 
Temperatures greater than 34°C 
Difference between maximum and minimum 
Standard deviation 
 
"""

temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])

print("Mean temperature:", temperature.mean())
print("Maximum temperature:", temperature.max())
print("Minimum temperature:", temperature.min())
print("Number of days above average:", np.sum(temperature > temperature.mean()))
print("Temperatures greater than 34°C:", temperature[temperature > 34])
print("Difference between maximum and minimum:", temperature.max() - temperature.min())
print("Standard deviation:", temperature.std())
```

```python
""" 
13. Monthly sales of a company are: 
sales = np.array([ 
    12000, 15000, 17000, 14000, 
    18000, 21000, 19000, 22000, 
    25000, 23000, 26000, 28000 
]) 
 
Determine: 
 
Total annual sales 
Average monthly sales 
Month with the highest sales 
Month with the lowest sales 
Number of months with sales above ₹20,000 
Percentage increase from January to December 
 
Hint: Remember that NumPy indexing starts at 0, while month numbering 
normally starts at 1.
"""

sales = np.array([
    12000, 15000, 17000, 14000,
    18000, 21000, 19000, 22000,
    25000, 23000, 26000, 28000
])

print("Total annual sales:", sales.sum())
print("Average monthly sales:", sales.mean())   
print("Month with the highest sales:", np.argmax(sales) + 1)
print("Month with the lowest sales:", np.argmin(sales) + 1)
print("Number of months with sales above ₹20,000:", np.sum(sales > 20000))
print("Percentage increase from January to December:", (sales[-1] - sales[0]) / sales[0] * 100)
```

```python
""" 
14. Consider marks of five students in four subjects: 
 
marks = np.array([ 
    [78, 85, 90, 88], 
    [65, 70, 72, 68], 
    [92, 88, 95, 91], 
    [55, 60, 58, 62], 
    [80, 75, 82, 79] 
]) 
 
Write a NumPy program that calculates: 
 
Total marks of every student 
Average of every student 
Average mark for every subject 
Highest-scoring student 
Lowest-scoring student 
Students whose average is greater than 80 
Overall class average 
 
Then assign a performance category based on average: 
 
|  Average | Category          | 
| -------: | ----------------- | 
|     ≥ 85 | Excellent         | 
| 70–84.99 | Good              | 
| 50–69.99 | Average           | 
|     < 50 | Needs Improvement | 
 
Challenge: Try assigning these categories using np.where() or np.select() 
instead of a Python loop.
"""
marks = np.array([
    [78, 85, 90, 88],
    [65, 70, 72, 68],
    [92, 88, 95, 91],
    [55, 60, 58, 62],
    [80, 75, 82, 79]
])  
print("Total marks of every student:", marks.sum(axis=1))
print("Average of every student:", marks.mean(axis=1))  
print("Average mark for every subject:", marks.mean(axis=0))
highest_scoring_student = np.argmax(marks.sum(axis=1))
lowest_scoring_student = np.argmin(marks.sum(axis=1))
print("Highest-scoring student:", highest_scoring_student + 1)
print("Lowest-scoring student:", lowest_scoring_student + 1)
print("Students whose average is greater than 80:", np.where(marks.mean(axis=1) > 80)[0] + 1)
print("Overall class average:", marks.mean())

scores = marks.mean(axis=1)
conditions = [scores >= 85, (scores >= 70) & (scores < 85), (scores >= 50) & (scores < 70)]
categories = ['Excellent', 'Good', 'Average']

performance_categories = np.select(conditions, categories, default='Needs Improvement')
print("Performance categories:\n", performance_categories)
```

```python
""" 
21. Take your own example to normalize an array.  
"""

def normal(arr):
    normalized_arr = (arr - np.min(arr)) / (np.max(arr) - np.min(arr))
    return normalized_arr

x= np.array([10, 34,51,66,4,67])
normal(x)
```

```python
""" 
22. use your own data to standardize an array. 
"""


def standardize(arr):
    standardized_arr = (arr - np.mean(arr)) / np.std(arr)
    return standardized_arr

x = np.array([10, 34, 51, 66, 4, 67])
standardize(x)
```

```python
""" 
23. Create a NumPy array containing 100 random exam marks and determine 
how many students fall into the ranges 0–39, 40–59, 60–79, and 80–100. 
"""
exam_marks = np.random.randint(0, 101, size=100)
print("Random exam marks:\n", exam_marks)

# Count students in each range
count_0_39 = np.sum((exam_marks >= 0) & (exam_marks <= 39))
count_40_59 = np.sum((exam_marks >= 40) & (exam_marks <= 59))
count_60_79 = np.sum((exam_marks >= 60) & (exam_marks <= 79))
count_80_100 = np.sum((exam_marks >= 80) & (exam_marks <= 100))

print("Number of students in range 0–39:", count_0_39)
print("Number of students in range 40–59:", count_40_59)
print("Number of students in range 60–79:", count_60_79)
print("Number of students in range 80–100:", count_80_100)
```

```python
""" 
24. Compare the time required to multiply 1,000,000 numbers using a 
Python list and a NumPy array.
"""


# Create a Python list and a NumPy array with 1,000,000 numbers
python_list = list(range(1, 1000001))
numpy_array = np.arange(1, 1000001)

# Measure time for Python list multiplication
start_time = time.time()
result_list = [x * 2 for x in python_list]
end_time = time.time()
time_list = end_time - start_time

# Measure time for NumPy array multiplication
start_time = time.time()
result_array = numpy_array * 2
end_time = time.time()
time_array = end_time - start_time

print("Time required for Python list multiplication:", time_list)
print("Time required for NumPy array multiplication:", time_array)
print("Numpy is aproximately {:.2f} times faster than Python list.".format(time_list / time_array))
```
