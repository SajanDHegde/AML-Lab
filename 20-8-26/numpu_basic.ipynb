```python
# math functions
a = np.array([-4, -3, 5, 6, 6])
print("Sum :",a.sum()) # sum of array
print("Mean :",a.mean()) # mean of array
print("Max :",a.max()) # max of array
print("Min :",a.min()) # min of array
print("Absolute Value :",np.abs(a)) # absolute value
```

```python
# argmax and argmin return the index of max and min value in the array
print("Index of Max :",a.argmax()) # index of max value
print("Index of Min :",a.argmin()) # index of min value
```

```python
# standard deviation and variance
print("Standard Deviation :",a.std()) # standard deviation
print("Variance :",a.var()) # variance
```

```python
# sum, mean, max, min, std, var , argmax, argmin on the array b
b = np.array([7,3,-2,0,100,25])
print("Sum :",b.sum()) # sum of array
print("Mean :",b.mean()) # mean of array
print("Max :",b.max()) # max of array
print("Min :",b.min()) # min of array
print("Standard Deviation :",b.std()) # standard deviation
print("Variance :",b.var()) # variance
print("Index of Max :",b.argmax()) # index of max value
print("Index of Min :",b.argmin()) # index of min value
```

```python
# Indexing and slicing
s = np.arange(100)
s
```

```python
s[0], s[4], s[25], s[99]
# use bracket notation to get the index of the array
```

```python
# use slicing to get the index of the array
print("first 10 elements:",s[0:10]) # first 10 elements
print("last 10 elements:",s[90:100]) # last 10 elements
print("every other element:",s[::2]) # every other element
print("every other element starting from index 1:",s[1::2]) # every other element starting from index 1
print("reversed array:",s[::-1]) # reverse the array
print("every other element from index 10 to 20:",s[10:20:2]) # every other element from index 10 to 20
```
